```python

condicao = df["smoker"] == 'yes'
df_filtrado = df[condicao]

df_agregado = df_filtrado.groupby(["sex"])["charges"].sum()
df_agregado

# rc = runtime configuration (configuração quando executa)
cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)
 
plt.bar(df_agregado.index, df_agregado.values)
# plt.bar(df_agregado.index[1], df_agregado["male"], bottom=df_agregado["male"], label='Homens')

# # forçar os anos a serem exibidos como inteiros
# plt.xticks(ticks=anos)

# plt.xlabel('Ano')
plt.ylabel('Gasto com Saude em Fumantes (em milhões)')
plt.title('Gasto com Saude em Fumantes (em milhões) por sexo')
# plt.legend()
plt.show()
```
